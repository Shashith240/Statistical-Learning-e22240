## Q. Statistical Mechanics of the Quantum Oscillator

### 1. The Partition Function

The canonical partition function is defined as:


$$Z(T) = \sum_{j=0}^{\infty} e^{-E_j / (k_B T)}$$

Given the energy eigenvalues $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$, we substitute this into the summation:


$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{\hbar \omega (j + 1/2)}{k_B T}} = \sum_{j=0}^{\infty} e^{-\frac{\hbar \omega}{2k_B T}} \cdot \left(e^{-\frac{\hbar \omega}{k_B T}}\right)^j$$

We can pull out the constant factor that does not depend on $j$:


$$Z(T) = e^{-\frac{\hbar \omega}{2k_B T}} \sum_{j=0}^{\infty} \left(e^{-\frac{\hbar \omega}{k_B T}}\right)^j$$

The summation is an infinite geometric series of the form $\sum_{j=0}^{\infty} r^j = \frac{1}{1-r}$, where the common ratio is $r = e^{-\frac{\hbar \omega}{k_B T}}$. Since $T > 0$, $|r| < 1$, the series converges to:


$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

---

### 2. Thermodynamic Potentials

#### **Helmholtz Free Energy ($F$)**

Using the relation $F(T) = -k_B T \ln Z$:


$$F(T) = -k_B T \ln \left( \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \right)$$

$$F(T) = -k_B T \left[ \ln\left(e^{-\frac{\hbar \omega}{2k_B T}}\right) - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

#### **Internal Energy ($U$)**

First, rewrite $\ln Z$ in terms of $\beta = \frac{1}{k_B T}$:


$$Z = \frac{e^{-\frac{\beta \hbar \omega}{2}}}{1 - e^{-\beta \hbar \omega}} \implies \ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$

Now take the negative derivative with respect to $\beta$:


$$U = -\frac{\partial \ln Z}{\partial \beta} = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot (-\hbar \omega) \right]$$

$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$$

Substituting back $\beta = \frac{1}{k_B T}$:


$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

### 3. Entropy

Using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)_V$:


$$S = -\frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar \omega}{k_B T}}\right) \cdot \left(\frac{\hbar \omega}{k_B T^2}\right)$$

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \frac{e^{-\frac{\hbar \omega}{k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)}$$

#### **Verification ($U = F + TS$)**

$$F + TS = \left[\frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)\right] + T \left[-k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)}\right]$$


The log terms cancel out completely:


$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U$$


Hence verified.

### 4. Heat Capacity and Hyperbolic Form

#### **Direct Calculation**

$$C_V = \frac{\partial U}{\partial T} = \frac{\partial}{\partial T} \left( \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right)$$

$$C_V = \hbar \omega \cdot \frac{-1}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} \cdot e^{\frac{\hbar \omega}{k_B T}} \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right) = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

#### **Hyperbolic Form Conversion**

Multiply the numerator and denominator by $e^{-\frac{\hbar \omega}{k_B T}}$:


$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(e^{\frac{\hbar \omega}{2k_B T}} - e^{-\frac{\hbar \omega}{2k_B T}}\right)^2}$$

Using the identity $\sinh(x) = \frac{e^x - e^{-x}}{2} \implies e^x - e^{-x} = 2\sinh(x)$, let $x = \frac{\hbar \omega}{2k_B T}$:


$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(2 \sinh\left(\frac{\hbar \omega}{2k_B T}\right)\right)^2}$$

$$C_V = k_B \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$

### 5. Asymptotic Behavior (Discussion)

In the high-temperature limit where $k_B T \gg \hbar \omega$, the parameter $x = \frac{\hbar \omega}{2k_B T} \to 0$.

Using the Taylor approximation for small values, $\sinh(x) \approx x$:


$$C_V \approx k_B (x)^2 \cdot \frac{1}{x^2} = k_B$$

This perfectly matches the classical equipartition theorem prediction for a 1D harmonic oscillator (which possesses two degrees of freedom: kinetic and potential, each contributing $\frac{1}{2}k_B$ to the heat capacity).

## Q. The Maximum Entropy Distribution under Mean and Variance Constraints

We maximize the differential entropy $H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) dx$ subject to three operational constraints using Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2$:

1. $\int_{-\infty}^{\infty} f(x) dx = 1$
2. $\int_{-\infty}^{\infty} x f(x) dx = \mu$
3. $\int_{-\infty}^{\infty} (x - \mu)^2 f(x) dx = \sigma^2$

The Lagrangian functional optimization yields the structural form:


$$f(x) = \exp\left(-\lambda_0 - 1 - \lambda_1 x - \lambda_2 (x - \mu)^2\right)$$

By redefining our constants to complete the square in the exponent, we can rewrite this as:


$$f(x) = C \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$

Using the normalization constraint to solve for the scaling constant $C$:


$$\int_{-\infty}^{\infty} C \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right) dx = 1 \implies C \cdot \sqrt{2\pi\sigma^2} = 1 \implies C = \frac{1}{\sqrt{2\pi\sigma^2}}$$

Thus, the probability density function uniquely matching maximum entropy is the Gaussian distribution:


$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$

## Q. Multiclass Logistic Regression: Automatic Classification of Emails

To explicitly determine the maximum entropy routing distribution, we define the conditional probabilities $P(Y=1|x)$ and $P(Y=0|x)$ using a Gibbs distribution framework. Let the probability of routing to the Technical team ($Y=1$) given feature state vector $x$ be:


$$P(Y=1|x) = \frac{\exp(\theta^T x)}{1 + \exp(\theta^T x)}$$

Where $\theta = (\theta_1, \theta_2, \theta_3)^T$ represents the vector of Lagrange parameters tied to the historical audit expectations. The expected empirical constraints match the model expectations:


$$E[x_i \cdot I(Y=1)] = \sum_{s \in \{A, B, C, D, E\}} P(x_s) \cdot x_{s,i} \cdot P(Y=1|x_s)$$

Using the values given for $P(x_s)$ and the observed dataset expectations, we set up our transcendental system of equations:

1. $0.40 \cdot P(Y=1|x_A) + 0.05 \cdot P(Y=1|x_E) = 0.05$
2. $0.25 \cdot P(Y=1|x_B) + 0.15 \cdot P(Y=1|x_D) + 0.05 \cdot P(Y=1|x_E) = 0.15$
3. $0.05 \cdot P(Y=1|x_C) + 0.15 \cdot P(Y=1|x_D) = 0.16$

Solving this system gives the conditional probabilities for each scenario, from which we retrieve the weight vectors via the log-odds relationship:


$$\theta_i = \ln \left( \frac{P(Y=1|x_i)}{1 - P(Y=1|x_i)} \right)$$

## Q. Asset Pricing and Portfolio Optimization

We seek to find $p(S_T)$ maximizing $H(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) dS_T$ under the given asset parameters:

1. $\int_{0}^{\infty} p(S_T) dS_T = 1$
2. $\int_{0}^{\infty} S_T p(S_T) dS_T = 102$
3. $\int_{0}^{\infty} S_T^2 p(S_T) dS_T = 10600$
4. $\int_{0}^{\infty} \ln(S_T) p(S_T) dS_T = 4.60$

The formal optimization template yields:


$$p(S_T) = \exp\left(-\lambda_0 - 1 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right)$$

$$p(S_T) = C \cdot S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

This family directly matches the **Generalized Gamma Distribution** profile. When down-crash risks are factored into portfolio modeling without enforcing symmetry ($\lambda_2 \neq 0$ and $\lambda_3 \neq 0$), the tail distributions natively decay faster or slower than a standard log-normal, precisely replicating the skewed implied volatility smirk seen on modern options exchanges.
