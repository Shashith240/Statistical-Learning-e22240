## Q. Analytical Derivation

### 1. Show that $x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$

Given the process model equation:
$$x_k^- = A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}$$

Taking the mathematical expectation (mean) on both sides:
$$m_k^- = \mathbb{E}[x_k^-] = \mathbb{E}[A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}]$$
$$m_k^- = A_{k-1} \mathbb{E}[x_{k-1}^+] + G_{k-1} \mathbb{E}[w_{k-1}]$$

Since $\mathbb{E}[x_{k-1}^+] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1} m_{k-1}$$

Now, compute the error vector to determine the covariance:
$$x_k^- - m_k^- = (A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}) - A_{k-1} m_{k-1} = A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1}$$

The covariance matrix $P_k^-$ is defined as:
$$P_k^- = \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right]$$
$$P_k^- = \mathbb{E}\left[\left(A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1}\right) \left(A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1} w_{k-1}\right)^T\right]$$

Since the state $x_{k-1}^+$ and process noise $w_{k-1}$ are mutually independent, their cross-expectation terms vanish ($\mathbb{E}[(x_{k-1}^+ - m_{k-1})w_{k-1}^T] = 0$). Expanding the product gives:
$$P_k^- = A_{k-1} \mathbb{E}\left[(x_{k-1}^+ - m_{k-1})(x_{k-1}^+ - m_{k-1})^T\right] A_{k-1}^T + G_{k-1} \mathbb{E}\left[w_{k-1} w_{k-1}^T\right] G_{k-1}^T$$
$$P_k^- = A_{k-1} P_{k-1} A_{k-1}^T + G_{k-1} \Sigma_p G_{k-1}^T$$

---

### 2. Show that $y_k^- \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$

Given the measurement equation:
$$y_k^- = H_k x_k^- + z_k$$

Taking the expectation:
$$\mathbb{E}[y_k^-] = \mathbb{E}[H_k x_k^- + z_k] = H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k]$$

Since $\mathbb{E}[x_k^-] = m_k^-$ and $\mathbb{E}[z_k] = 0$:
$$\mathbb{E}[y_k^-] = H_k m_k^-$$

Now, determine the measurement error vector:
$$y_k^- - \mathbb{E}[y_k^-] = H_k(x_k^- - m_k^-) + z_k$$

The covariance matrix is:
$$\text{Var}(y_k^-) = \mathbb{E}\left[(y_k^- - \mathbb{E}[y_k^-])(y_k^- - \mathbb{E}[y_k^-])^T\right]$$
$$\text{Var}(y_k^-) = \mathbb{E}\left[\left(H_k(x_k^- - m_k^-) + z_k\right)\left(H_k(x_k^- - m_k^-) + z_k\right)^T\right]$$

Since $x_k^-$ and $z_k$ are independent:
$$\text{Var}(y_k^-) = H_k \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] H_k^T + \mathbb{E}[z_k z_k^T]$$
$$\text{Var}(y_k^-) = H_k P_k^- H_k^T + \Sigma_m$$

---

### 3. Show the Joint Distribution of $x_k^-$ and $y_k^-$

To construct the joint Gaussian distribution, we compute the cross-covariance term:
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[(x_k^- - m_k^-)(y_k^- - \mathbb{E}[y_k^-])^T\right]$$
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[(x_k^- - m_k^-)\left(H_k(x_k^- - m_k^-) + z_k\right)^T\right]$$
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] H_k^T + \mathbb{E}[(x_k^- - m_k^-)z_k^T]$$

Using the independence of state and measurement noise, the second term evaluates to zero:
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$

By symmetry, $\text{Cov}(y_k^-, x_k^-) = H_k P_k^-$. Combining the means and covariances yields the joint distribution:
$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix} \right)$$

---

### 4. Show the Conditional Distribution Update ($x_k^+$)

Applying the standard Gaussian conditioning theorem to the joint distribution above, if:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left(\begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix}\right)$$

Then the conditional distribution $(X \mid Y = y)$ is Gaussian with:
$$\mu_{X|Y} = \mu_X + \Sigma_{XY} \Sigma_{YY}^{-1}(y - \mu_Y)$$
$$\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY} \Sigma_{YY}^{-1} \Sigma_{YX}$$

Substituting our joint terms ($X = x_k^-$, $Y = y_k^-$ and $y = y_k^{\text{obs}}$):
$$K_k = \Sigma_{XY} \Sigma_{YY}^{-1} = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

The updated mean becomes:
$$m_k = m_k^- + K_k (y_k^{\text{obs}} - H_k m_k^-)$$

The updated covariance becomes:
$$P_k = P_k^- - K_k H_k P_k^- = (I - K_k H_k) P_k^-$$

Thus, $x_k^+ \sim \mathcal{N}(m_k, P_k)$.

---

### 5. Find $\mathbb{E}[x_k^- \mid y_k^- = y_k^{\text{obs}}]$ and $\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}})$

By definition, the conditional expectation and conditional variance of the state given the observed measurement are precisely the updated filter state mean ($m_k$) and covariance ($P_k$) derived in part 4:

$$\mathbb{E}[x_k^- \mid y_k^- = y_k^{\text{obs}}] = m_k = m_k^- + K_k (y_k^{\text{obs}} - H_k m_k^-)$$
$$\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}}) = P_k = (I - K_k H_k) P_k^-$$

## Q. 1-D Example

### 1. Show $m_k^- = a m_{k-1}$ and $P_k^- = a^2 P_{k-1} + q$
Using the general formulas with 1D scalars ($A_{k-1}=a$, $G_{k-1}=1$, $\Sigma_p=q$):
$$m_k^- = A_{k-1} m_{k-1} = a m_{k-1}$$
$$P_k^- = A_{k-1} P_{k-1} A_{k-1}^T + G_{k-1} \Sigma_p G_{k-1}^T = a P_{k-1} a + (1) q (1) = a^2 P_{k-1} + q$$

### 2. Show the update equations for $m_k$ and $P_k$
With scalar values $H_k = h$, $\Sigma_m = r$, the innovation variance is $S_k = h^2 P_k^- + r$.
The Kalman gain simplifies to:
$$K_k = \frac{P_k^- h}{S_k}$$

Thus:
$$m_k = m_k^- + K_k(y_k^{\text{obs}} - h m_k^-) = m_k^- + \frac{P_k^- h}{S_k}(y_k^{\text{obs}} - h m_k^-)$$
$$P_k = (1 - K_k h) P_k^- = \left(1 - \frac{P_k^- h^2}{S_k}\right) P_k^-$$

### 3. Show $p(y_k^- \mid Y_{k-1}) = \mathcal{N}(h m_k^-, h^2 P_k^- + r)$
This is the distribution of the measurement prior to evaluating the current observation, which matches the formulation derived in section 1 part 2 under scalar substitutions.

### 4. Show $p(y_k^- \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$
Once the measurement $y_k$ is absorbed to update the state to $x_k^+ \sim \mathcal{N}(m_k, P_k)$, any post-fit diagnostic calculation or posterior representation follows the same mapping:
$$\mathbb{E}[y_k^- \mid Y_k] = h \mathbb{E}[x_k^+ \mid Y_k] = h m_k$$
$$\text{Var}(y_k^- \mid Y_k] = h^2 \text{Var}(x_k^+ \mid Y_k) + r = h^2 P_k + r$$

### 5. Numerical Python Animation Code Block


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.stats import norm
from IPython.display import HTML

# Parameters
a, q, h, r = 0.9, 0.4, 1.0, 0.2
m0, P0 = 0.0, 1.0
np.random.seed(42)

# Generate synthetic trajectory
steps = 30
true_x = np.zeros(steps)
obs_y = np.zeros(steps)
x_val = m0

for k in range(steps):
    x_val = a * x_val + np.random.normal(0, np.sqrt(q))
    true_x[k] = x_val
    obs_y[k] = h * x_val + np.random.normal(0, np.sqrt(r))

# Arrays to capture distributions
history = []
m, P = m0, P0

for k in range(steps):
    # Predict
    m_minus = a * m
    P_minus = a**2 * P + q
    # Update
    S = h**2 * P_minus + r
    K = (P_minus * h) / S
    m = m_minus + K * (obs_y[k] - h * m_minus)
    P = (1 - K * h) * P_minus
    history.append((m_minus, P_minus, m, P))

# Animation Setup
fig, ax = plt.subplots(figsize=(10, 5))
x_axis = np.linspace(-5, 5, 500)

def update_frame(k):
    ax.clear()
    m_m, P_m, m_u, P_u = history[k]
    prior_pdf = norm.pdf(x_axis, m_m, np.sqrt(P_m))
    post_pdf = norm.pdf(x_axis, m_u, np.sqrt(P_u))
    
    ax.plot(x_axis, prior_pdf, 'r--', label=f'Prior Predictive $\mathcal{{N}}$({m_m:.2f}, {P_m:.2f})')
    ax.plot(x_axis, post_pdf, 'b-', label=f'Posterior Filtered $\mathcal{{N}}$({m_u:.2f}, {P_u:.2f})')
    ax.axvline(true_x[k], color='green', linestyle='-', linewidth=2, label=f'True State: {true_x[k]:.2f}')
    ax.axvline(obs_y[k], color='orange', linestyle=':', linewidth=2, label=f'Observation: {obs_y[k]:.2f}')
    
    ax.set_title(f"Time Step k = {k+1}")
    ax.set_ylim(0, 1.5)
    ax.legend(loc='upper right')
    ax.grid(True)

anim = FuncAnimation(fig, update_frame, frames=steps, interval=600)
plt.close()
HTML(anim.to_jshtml())
