## Part 1: Derivation of the Sample Covariance Bias and Bessel's Correction

To understand why the Maximum Likelihood Estimator (MLE) for the covariance matrix is biased, we first state the definition of the raw MLE estimator:

$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

We aim to find its expected value $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]$.

---

### Step 1: Decomposing the Deviation Term
Let us algebraicly manipulate the term $(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)$ by adding and subtracting the true population mean $\boldsymbol{\mu}$:

$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Taking the outer product of this vector with itself yields:

$$(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the right-hand side:

$$= (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

---

### Step 2: Summing Over All Observations
Now, we sum this expression from $i=1$ to $n$:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Notice that $(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$ does not depend on the summation index $i$. Furthermore, observe that:

$$\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = \left(\sum_{i=1}^n \mathbf{X}_i\right) - n\boldsymbol{\mu} = n\widehat{\boldsymbol{\mu}}_n - n\boldsymbol{\mu} = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Using this identity, the cross-product terms simplify cleanly:

$$\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T = \left[ \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right] (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Similarly, the third term simplifies to the exact same expression. The fourth term is a sum of a constant matrix over $n$ steps:

$$\sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Substituting these back into our total sum:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

---

### Step 3: Evaluating Expectations
We take the expectation of both sides. By definition of the population covariance $\boldsymbol{\Sigma}$:

$$\mathbb{E}\left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] = \boldsymbol{\Sigma}$$

Thus, the first term on the right becomes:

$$\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] = \sum_{i=1}^n \boldsymbol{\Sigma} = n\boldsymbol{\Sigma}$$

For the second term, we evaluate the covariance of the sample mean $\widehat{\boldsymbol{\mu}}_n$:

$$\mathbb{E}\left[ (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right] = \mathrm{Var}(\widehat{\boldsymbol{\mu}}_n) = \mathrm{Var}\left( \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i \right) = \frac{1}{n^2} \sum_{i=1}^n \mathrm{Var}(\mathbf{X}_i) = \frac{1}{n^2} (n\boldsymbol{\Sigma}) = \frac{1}{n}\boldsymbol{\Sigma}$$

Multiplying this by the leading factor of $n$:

$$\mathbb{E}\left[ n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right] = n \left( \frac{1}{n}\boldsymbol{\Sigma} \right) = \boldsymbol{\Sigma}$$

Combining these results gives the expectation of our sum:

$$\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = n\boldsymbol{\Sigma} - \boldsymbol{\Sigma} = (n-1)\boldsymbol{\Sigma}$$

---

### Step 4: Proving Finite-Sample Bias and Unbiased Correction

Now we apply this expectation to the raw Maximum Likelihood Estimator:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \mathbb{E}\left[ \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \frac{1}{n} (n-1)\boldsymbol{\Sigma} = \frac{n-1}{n}\boldsymbol{\Sigma}$$

> **Conclusion on Bias:** Because $\frac{n-1}{n} \neq 1$ for any finite sample size $n$, $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ systematically underestimates the true population covariance. The bias arises because the sample mean $\widehat{\boldsymbol{\mu}}_n$ is dynamically closer to the observed sample cluster than the true population mean $\boldsymbol{\mu}$ is, burning up exactly one degree of freedom.

To construct an **unbiased** estimator $\mathbf{S}$, we isolate $\boldsymbol{\Sigma}$ by scaling the expression by its reciprocal factor $\frac{n}{n-1}$:

$$\boldsymbol{\Sigma} = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]$$

$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

By linearity of expectation:

$$\mathbb{E}[\mathbf{S}] = \mathbb{E}\left[ \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} \right] = \frac{n}{n-1} \left( \frac{n-1}{n}\boldsymbol{\Sigma} \right) = \boldsymbol{\Sigma}$$

This proves that applying Bessel's correction ($\frac{1}{n-1}$) yields a unique, unbiased estimator.

## Part 2: Statistical Errors and Geometry in Structural Health Monitoring

### 1. Error Definitions in Structural Health Monitoring
* **Type I Error ($\alpha$):** The probability of committing a False Alarm. Formally, it occurs when the engineering diagnostic framework rejects the null hypothesis $H_0$ (concluding the asset is structurally damaged) when in reality $H_0$ is true (the structural asset is completely healthy).
    $$\alpha = P(\text{Reject } H_0 \mid H_0 \text{ is true})$$
* **Type II Error ($\beta$):** The probability of a Missed Detection. Formally, it occurs when the framework fails to reject $H_0$ (concluding the asset is running safely) when in reality $H_0$ is false (the asset has suffered structural damage).
    $$\beta = P(\text{Fail to Reject } H_0 \mid H_0 \text{ is false})$$

---

### 2. Mathematical Consequences of Ultra-Conservative Thresholds ($\alpha = 0.0001$)

#### Effect on Statistical Power
In statistical hypothesis testing, the Type I error probability $\alpha$ and the Type II error probability $\beta$ are inversely linked given a fixed sample size $n$. When an engineer dramatically restricts the false alarm threshold to an ultra-conservative limit ($\alpha \to 0.0001$), the critical acceptance region under the null distribution expands significantly to capture almost all random variations.

Consequently, the threshold for rejecting $H_0$ moves deep into the tails of the distribution. This directly inflates the probability of missing true underlying shifts, causing $\beta$ to spike.

Because the statistical power of a diagnostic system is defined as:

$$\text{Power} = 1 - \beta$$

an ultra-low $\alpha$ directly suppresses the statistical power. The diagnostic system becomes largely blind to subtle, early-stage structural degradation, meaning an industrial failure could occur without triggering an alarm.

#### Geometric Volume modification of the Confidence Ellipsoid
For multivariate normal asset observations where $\mathbf{X} \sim \mathcal{N}_m(\boldsymbol{\mu}, \boldsymbol{\Sigma})$, the boundary of the healthy operation domain is governed by the Mahalanobis distance, creating a hyper-ellipsoid defined by:

$$(\mathbf{X} - \widehat{\boldsymbol{\mu}}_n)^T \mathbf{S}^{-1} (\mathbf{X} - \widehat{\boldsymbol{\mu}}_n) \le \chi^2_{m}(\alpha)$$

where $\chi^2_{m}(\alpha)$ is the critical value of the Chi-squared distribution with $m$ degrees of freedom at significance level $\alpha$.

## 1. Rigorous Statement of Slutsky’s Theorem

Slutsky’s Theorem describes the asymptotic behavior of algebraic combinations of sequences of random variables (or vectors) where one sequence converges in distribution and the other converges in probability to a constant.

### Multivariate Formulation
Let $\mathbf{X}_n$ be a sequence of random $k$-dimensional vectors, and let $\mathbf{Y}_n$ be a sequence of random $k \times k$ matrices. Let $\mathbf{X}$ be a random $k$-dimensional vector, and let $\mathbf{C}$ be a deterministic, constant $k \times k$ matrix.

If:
1. $\mathbf{X}_n \xrightarrow{d} \mathbf{X}$ (Convergence in distribution)
2. $\mathbf{Y}_n \xrightarrow{p} \mathbf{C}$ (Convergence in probability)

Then:
* **Addition:** $\mathbf{X}_n + \mathbf{Y}_n\mathbf{v} \xrightarrow{d} \mathbf{X} + \mathbf{C}\mathbf{v}$ (for any constant vector $\mathbf{v}$)
* **Multiplication:** $\mathbf{Y}_n \mathbf{X}_n \xrightarrow{d} \mathbf{C} \mathbf{X}$
* **Inverse Multiplication:** $\mathbf{Y}_n^{-1} \mathbf{X}_n \xrightarrow{d} \mathbf{C}^{-1} \mathbf{X}$ (provided $\mathbf{C}$ is non-singular and $\mathbf{Y}_n^{-1}$ exists infinitely often)

---

## 2. Proof of Operationalizing the Empirical Estimator

We want to justify substituting the empirical sample covariance matrix $\mathbf{S}_n$ for the latent population covariance matrix $\boldsymbol{\Sigma}$, transforming our theoretical asymptotic result into a usable frequentist model.

### Step 1: Establish the Asymptotic Benchmarks
From the Lindeberg-Lévy Multivariate Central Limit Theorem (CLT), we have:

$$\mathbf{Z}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathbf{Z} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

By the Weak Law of Large Numbers (WLLN) or the Continuous Mapping Theorem applied to sample moments, the empirical sample covariance matrix $\mathbf{S}_n$ is a consistent estimator of the population covariance matrix $\boldsymbol{\Sigma}$:

$$\mathbf{S}_n \xrightarrow{p} \boldsymbol{\Sigma} \quad \text{as } n \to \infty$$

Because the matrix square root and matrix inversion are continuous operations on the space of positive-definite matrices, the Continuous Mapping Theorem yields:

$$\mathbf{S}_n^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

### Step 2: Application of Slutsky's Theorem
We now construct a standardized pivot sequence. Let $\mathbf{X}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$ and $\mathbf{Y}_n = \mathbf{S}_n^{-1/2}$.

Applying the multiplication rule of Slutsky's Theorem ($\mathbf{Y}_n \mathbf{X}_n \xrightarrow{d} \mathbf{C} \mathbf{X}$):

$$\mathbf{S}_n^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathbf{Z}$$

where $\mathbf{Z} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$.

### Step 3: Distribution of the Limiting Variable
To find the explicit distribution of the right-hand side, we compute the linear transformation of the multivariate normal vector $\mathbf{Z}$:

$$\mathbb{E}[\boldsymbol{\Sigma}^{-1/2} \mathbf{Z}] = \boldsymbol{\Sigma}^{-1/2} \mathbb{E}[\mathbf{Z}] = \mathbf{0}$$

$$\text{Var}(\boldsymbol{\Sigma}^{-1/2} \mathbf{Z}) = \boldsymbol{\Sigma}^{-1/2} \text{Var}(\mathbf{Z}) \left(\boldsymbol{\Sigma}^{-1/2}\right)^T = \boldsymbol{\Sigma}^{-1/2} \boldsymbol{\Sigma} \boldsymbol{\Sigma}^{-1/2} = \mathbf{I}_k$$

Thus, the standardized sequence converges to a standard multivariate Gaussian distribution:

$$\mathbf{S}_n^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I}_k)$$

### Step 4: Practical Parametric Modeling Approximation
For a fixed, sufficiently large sample size $n$, we can invert the standardization transformation. By scaling back by $\mathbf{S}_n^{1/2}$, we obtain:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \approx \mathscr{N}(\mathbf{0}, \mathbf{S}_n)$$

Dividing the system by $\sqrt{n}$ distributes the variance component by a factor of $(\frac{1}{\sqrt{n}})^2 = \frac{1}{n}$:

$$\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu} \approx \mathscr{N}\left(\mathbf{0}, \frac{1}{n}\mathbf{S}_n\right)$$

Adding the deterministic vector $\boldsymbol{\mu}$ yields the final practical parametric modeling distribution used in multivariate inference:

$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left({\boldsymbol{\mu}}, \frac{1}{n}\mathbf{S}_n\right)$$

> **Conclusion:** Slutsky's Theorem guarantees that substituting $\mathbf{S}_n$ for $\boldsymbol{\Sigma}$ introduces an error that vanishes in probability as $n \to \infty$, preserving the validity of the Gaussian approximation for inference tasks (such as constructing Wald-type confidence regions or Hotelling's $T^2$ tests).

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    X = df.to_numpy()
    n, m = X.shape
    n_j = n // g_chunks

    # 1. Compute Global Mean
    global_mean = np.mean(X, axis=0)

    # Initialize Matrices
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # 2. Partition into g blocks and compute W and B matrices
    for j in range(g_chunks):
        chunk = X[j * n_j : (j + 1) * n_j]
        chunk_mean = np.mean(chunk, axis=0)

        # Within-chunk sum of squares and cross-products (Error)
        chunk_deviations = chunk - chunk_mean
        W += chunk_deviations.T @ chunk_deviations

        # Between-chunk sum of squares and cross-products (Hypothesis)
        mean_diff = (chunk_mean - global_mean).reshape(-index, 1) if hasattr(chunk_mean, 'reshape') else np.expand_dims(chunk_mean - global_mean, axis=1)
        B += n_j * (mean_diff @ mean_diff.T)

    # 3. Compute Wilks' Lambda
    det_W = np.linalg.det(W)
    det_Total = np.linalg.det(B + W)
    wilks_lambda = det_W / det_Total

    # 4. Bartlett’s Chi-Square Approximation
    multiplier = (n - 1 - (m + g_chunks) / 2.0)
    chi2_calc = -multiplier * np.log(wilks_lambda)

    # Degrees of Freedom: m * (g - 1)
    dof = m * (g_chunks - 1)

    # Calculate p-value
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, dof)

    return {
        "Wilks' Lambda": wilks_lambda,
        "Bartlett Chi2": chi2_calc,
        "Degrees of Freedom": dof,
        "p-value": p_value
    }

# Execute your completed function
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

print("--- MANOVA / Bartlett First-Moment Homogeneity Report ---")
for key, val in results.items():
    print(f"{key:<20}: {val:.6f}" if isinstance(val, float) else f"{key:<20}: {val}")
````

## Part 1: Principal Component Transformation & De-correlation

### 1.1 Step-by-Step Proof of Covariance Diagonalization

To find the theoretical covariance matrix of the transformed vector $\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$, we evaluate its expected outer product. Since $\mathbb{E}[\widetilde{\mathbf{X}}_i] = \mathbf{0}$, it follows that $\mathbb{E}[\mathbf{Z}_i] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i] = \mathbf{0}$.

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right) \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right)^T \right]$$

Using the matrix transposition property $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P} \right]$$

Because $\mathbf{P}$ is a deterministic matrix derived from the population covariance spectral decomposition, it can be pulled outside the expectation operator ($\mathbb{E}$):

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}\left[ \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \right] \mathbf{P}$$

Substituting the baseline population covariance matrix $\boldsymbol{\Sigma} = \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T]$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Now, substitute the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \left( \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T \right) \mathbf{P} = \left(\mathbf{P}^T \mathbf{P}\right) \mathbf{\Lambda} \left(\mathbf{P}^T \mathbf{P}\right)$$

Since $\mathbf{P}$ is an orthogonal matrix composed of orthonormal eigenvectors, it satisfies $\mathbf{P}^T \mathbf{P} = \mathbf{I}_m$. Substituting the identity matrix yields:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I}_m \mathbf{\Lambda} \mathbf{I}_m = \mathbf{\Lambda}$$

$\blacksquare$

---

### 1.2 Geometric and Statistical Interpretation



* **Statistical Meaning:** The off-diagonal entries of the transformed covariance matrix represent the cross-covariances between distinct coordinates:
    $$\text{Cov}(Z_{i,j}, Z_{i,k}) = 0 \quad \forall j \neq k$$
    This proves that the transformation completely **de-correlates** the features. In the latent space, knowing the value of one principal component provides zero linear information about another. The diagonal entries $\lambda_j$ represent the variances of these independent components.
* **Geometric Meaning:** The orthogonal matrix $\mathbf{P}^T$ performs a rigid **rigid rotation** (and potential reflection) of the coordinate axes. It aligns the new coordinate axes $(\mathbf{Z})$ directly with the hyper-ellipsoid axes of the data's probability distribution. The cross-covariances vanish because the new axes are chosen precisely along the directions of maximum orthogonal variance.

## Part 2: Total Variance Conservation & Subspace Partitioning

### 2.1 Proof of Total Variance Conservation

The cyclic permutation property of the trace operator states that for any matrices $\mathbf{A}$ and $\mathbf{B}$ such that the products are square, $\text{tr}(\mathbf{A}\mathbf{B}) = \text{tr}(\mathbf{B}\mathbf{A})$.

We begin with the trace of the raw population covariance matrix:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T)$$

Group the matrices as $\mathbf{A} = \mathbf{P}$ and $\mathbf{B} = \mathbf{\Lambda} \mathbf{P}^T$. Applying the cyclic permutation rule:
$$\text{tr}\left( \mathbf{P} (\mathbf{\Lambda} \mathbf{P}^T) \right) = \text{tr}\left( (\mathbf{\Lambda} \mathbf{P}^T) \mathbf{P} \right)$$

Using matrix associativity and the orthogonality property $\mathbf{P}^T \mathbf{P} = \mathbf{I}_m$:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}\left( \mathbf{\Lambda} (\mathbf{P}^T \mathbf{P}) \right) = \text{tr}(\mathbf{\Lambda} \mathbf{I}_m) = \text{tr}(\mathbf{\Lambda})$$

Since $\mathbf{\Lambda}$ is a diagonal matrix containing the eigenvalues $\lambda_j$, its trace is simply the sum of its diagonal elements:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j$$

$\blacksquare$

---

### 2.2 Mathematical Formulations for Subspace Ratios

Given empirical eigenvalues sorted in descending order ($\hat{\lambda}_1 \ge \hat{\lambda}_2 \ge \dots \ge \hat{\lambda}_m$), the ratios for a truncation threshold $k$ are defined as:

* **Cumulative Explained Variance Ratio $\Phi(k)$:** Measures the proportion of total systemic variance retained within the principal subspace.
$$\Phi(k) = \frac{\text{tr}\left(\widehat{\mathbf{P}}_k^T \widehat{\boldsymbol{\Sigma}} \widehat{\mathbf{P}}_k\right)}{\text{tr}(\widehat{\boldsymbol{\Sigma}})} = \frac{\sum_{j=1}^k \hat{\lambda}_j}{\sum_{j=1}^m \hat{\lambda}_j}$$

* **Residual Unexplained Variance Ratio $\Psi(k)$:** Measures the remaining proportion of variance discarded into the residual subspace.
$$\Psi(k) = \frac{\text{tr}\left(\widehat{\mathbf{P}}_{m-k}^T \widehat{\boldsymbol{\Sigma}} \widehat{\mathbf{P}}_{m-k}\right)}{\text{tr}(\widehat{\boldsymbol{\Sigma}})} = \frac{\sum_{j=k+1}^m \hat{\lambda}_j}{\sum_{j=1}^m \hat{\lambda}_j} = 1 - \Phi(k)$$

## Part 3: Subspace Decomposition & Diagnostics in SHM

### 3.1 Vector Definitions and Energy Decomposition Proof

We define the components of the partitioned coordinate system as:
* **Reconstructed Vector:** $\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$
* **Residual Error Vector:** $\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$

#### Pythagorean Proof:
We express the squared Euclidean norm of the centered observation vector as an inner product:
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)$$

Substitute the partitioned representation $(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$:
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)$$

Expanding the terms via distribution:
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} + 2 \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

Because the full transformation matrix $\widehat{\mathbf{P}}$ is orthogonal, its sub-blocks are mutually orthogonal and individually orthonormal:
1. $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_k$
2. $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{m-k}$
3. $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}_{k \times (m-k)}$

Applying these properties simplifies the expression:
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \mathbf{I}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{I}_{m-k} \mathbf{z}_{i,m-k} + 0$$
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k}$$

Using the definition of the squared Euclidean norm ($\|\mathbf{v}\|^2 = \mathbf{v}^T\mathbf{v}$):
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

$\blacksquare$

### 3.2 Diagnostic Functions: $T^2$ versus $Q$ Statistic

| Diagnostic Metric | Mathematical Definition | Subspace Monitored | Functional Purpose |
| :--- | :--- | :--- | :--- |
| **Hotelling’s $T^2$** | $T^2_i = \mathbf{z}_{i,k}^T \mathbf{\Lambda}_k^{-1} \mathbf{z}_{i,k}$ | Principal (Explained) Subspace | Monitors shifts **within** the expected normal operating model boundary. |
| **Residual $Q$ (SPE)** | $Q_i = \|\mathbf{e}_i\|^2 = \|\mathbf{z}_{i,m-k}\|^2$ | Residual (Unexplained) Subspace | Monitors breakdowns **outside** the model space, indicating a violation of underlying correlation structures. |

#### Anomaly Event Matching & Justification

##### 1. Environmental Load Anomaly (e.g., Extreme Wind) $\rightarrow$ **Hotelling's $T^2$ Sensitivity**
* **Justification:** Extreme wind loads change the *amplitude* of structural vibrations but do not alter the underlying physical properties of the structure. The cross-variable correlations built during baseline conditions remain valid. Because the system state moves along known, expected correlation pathways to an extreme degree, this inflation of energy is captured inside the primary principal components, driving up the Mahalanobis distance measured by $T^2$.

##### 2. Internal Structural Fracture (e.g., Fatigue Crack) $\rightarrow$ **Residual $Q$ Statistic (SPE) Sensitivity**
* **Justification:** A structural fracture alters the physical load paths and boundary conditions of the system. This alters the underlying physical relationship between sensors, rendering the baseline covariance structure $\boldsymbol{\Sigma}$ invalid. Because the new behavior cannot be linearly represented by the trained principal axes $\widehat{\mathbf{P}}_k$, the projection error grows. The unaccounted energy spills directly into the residual subspace, making the $Q$ statistic the most sensitive indicator for this type of structural degradation.

## **Part B: Numerical Verification**



***Scenario:** You are building a real-time diagnostics layer for a structural component monitored by $m = 4$ sensors. You have a baseline archive of $n = 3,000$ nominal snapshots. To establish operational boundaries, you must write an optimization pipeline that analyzes how Hotelling’s $T^2$ and the residual $Q$ statistic respond to different truncation boundaries $k \in \{1, 2, 3\}$.*

**Your Task:**
Complete the Python class method below inside Google Colab. The function must:

1. Standardize and center the incoming realizations.
2. Decompose the unbiased sample covariance matrix $\mathbf{S}$ using `np.linalg.eigh` and sort components in descending order ($\widehat{\lambda}_1 \ge \dots \ge \widehat{\lambda}_m$).
3. Loop through every possible subspace cutoff value $k \in \{1, 2, 3\}$, calculating the sample vector array for Hotelling’s $T^2$ ($T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$) and the $Q$ statistic ($Q_i = \sum_{j=k+1}^m z_{i,j}^2$).
4. Compute the expected value (empirical mean) of $T^2$ and $Q$ across all samples for each $k$.
5. Return these metrics to verify your theoretical derivations in Part A.



````python
import numpy as np

class StructuralDiagnostics:
    def __init__(self, baseline_data: np.ndarray):
        """
        Initializes the diagnostics layer.
        :param baseline_data: np.ndarray of shape (n, m) where n=3000, m=4
        """
        self.X = baseline_data
        self.n, self.m = baseline_data.shape

    def compute_subspace_metrics(self) -> dict:
        """
        Standardizes data, decomposes the unbiased covariance matrix, and
        evaluates empirical means of T^2 and Q statistics for k in {1, 2, 3}.

        :return: dict containing empirical means of T^2 and Q for each k.
        """
        # 1. Standardize and center the incoming realizations (mean=0, std=1)
        # Using ddof=1 for sample standard deviation to match unbiased assumptions
        mean = np.mean(self.X, axis=0)
        std = np.std(self.X, axis=0, ddof=1)
        X_scaled = (self.X - mean) / std

        # 2. Decompose the unbiased sample covariance matrix S
        # S = (1 / (n - 1)) * X_scaled^T * X_scaled
        S = np.cov(X_scaled, rowvar=False)

        # np.linalg.eigh returns eigenvalues in ascending order for symmetric matrices
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort components in descending order
        idx = np.argsort(eigenvalues)[::-1]
        lambdas = eigenvalues[idx]
        V = eigenvectors[:, idx]

        # Transform scaled data into the principal component space: Z = X_scaled * V
        # Z has shape (n, m), where each column j corresponds to scores of PC_j
        Z = X_scaled @ V

        # Initialize dictionary to store the empirical means
        results = {}

        # 3. Loop through every possible subspace cutoff value k in {1, 2, 3}
        for k in [1, 2, 3]:
            # --- Hotelling's T^2 Statistic ---
            # T^2_i = sum_{j=1}^k (z_{i,j}^2 / lambda_j)
            # Vectorized across all n samples:
            T2_samples = np.sum((Z[:, :k] ** 2) / lambdas[:k], axis=1)

            # --- Residual Q Statistic ---
            # Q_i = sum_{j=k+1}^m (z_{i,j}^2)
            # Vectorized across all n samples:
            Q_samples = np.sum(Z[:, k:] ** 2, axis=1)

            # 4. Compute the expected value (empirical mean) across all samples
            mean_T2 = np.mean(T2_samples)
            mean_Q = np.mean(Q_samples)

            # 5. Store metrics
            results[k] = {
                "mean_T2": mean_T2,
                "mean_Q": mean_Q
            }

        return results

# ==========================================
# Verification & Execution Block
# ==========================================
if __name__ == "__main__":
    # Generate synthetic baseline data mimicking 4 sensors, 3000 snapshots
    np.random.seed(42)
    synthetic_snapshots = np.random.normal(loc=10.0, scale=2.5, size=(3000, 4))

    # Instantiate class and run pipeline
    diagnostics = StructuralDiagnostics(synthetic_snapshots)
    metrics = diagnostics.compute_subspace_metrics()

    # Print formatted results to verify against Part A theoretical derivations
    print(f"{'k':<5} | {'Empirical E[T²]':<18} | {'Empirical E[Q]':<18}")
    print("-" * 45)
    for k, val in metrics.items():
        print(f"{k:<5} | {val['mean_T2']:<18.4f} | {val['mean_Q']:<18.4f}")
````

## Part 1: The Fundamental Equation, Communality, and Uniqueness

### 1.1 Proof of the Fundamental Equation of Factor Analysis
We are given the linear generative model for a normalized, zero-mean baseline feature vector $\mathbf{Z}_i \in \mathbb{R}^m$:
$$\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$$

The population covariance matrix $\mathbf{R}$ is defined as the expected value of the outer product of $\mathbf{Z}_i$ with itself:
$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$$

Substituting the generative structure into the covariance definition:
$$\mathbf{R} = \mathbb{E}\left[ (\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T \right]$$

Expanding the outer product using the linearity of expectation:
$$\mathbf{R} = \mathbb{E}\left[ \boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T + \boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T \right]$$
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Using the model assumptions and constraints:
1. Orthogonal common factors: $\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}_{k \times k}$
2. Independent latent factors and errors: $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}$ (which also implies $\mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] = \mathbf{0}^T = \mathbf{0}$)
3. Diagonalized measurement noise: $\mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$

Substituting these identities back into the expanded equation yields:
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbf{I}_{k \times k} \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{0} + \mathbf{0} \boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$
$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

---

### 1.2 Diagonal Decomposition and Asset Diagnostics Interpretation
Focusing on the $j$-th diagonal entry of $\mathbf{R}$, the variance of the $j$-th feature (which equals $1$ because $\mathbf{Z}_i$ is normalized) is decomposed as:
$$R_{jj} = 1 = \sum_{l=1}^k \lambda_{jl}^2 + \varphi_j^2$$

#### Definitions
* **Communality ($h_j^2$):** The portion of the variance of the $j$-th feature explained by the $k$ shared latent factors.
    $$h_j^2 = \sum_{l=1}^k \lambda_{jl}^2$$
* **Uniqueness / Specific Variance ($\varphi_j^2$):** The portion of the variance specific to the $j$-th feature that cannot be explained by the common factors, representing localized measurement noise or unique sensor dynamics.
    $$\varphi_j^2 = 1 - h_j^2$$

#### Physical Interpretation in an Asset Diagnostics System
In a complex asset diagnostics framework (e.g., monitoring a gas turbine or industrial pump using an array of sensors):

* **Communality ($h_j^2$)** represents the degree to which a specific sensor's readings are driven by **global, systemic operational states** (such as core thermal load, rotational speed, or system pressure). A high communality means the sensor is highly integrated into the baseline physics of the machine; its variations capture systemic health or degradation profiles shared across multiple subsystems.
* **Uniqueness ($\varphi_j^2$)** represents the variance isolated to that **specific physical node**. It captures localized anomalies, electrical high-frequency noise, or sensor degradation (e.g., a loose thermocouple coupling or a localized seal leak). If a sensor suddenly exhibits an increase in unexplained residual energy, it indicates an isolated component fault rather than a global operational shift.

## Part 2: Factor Rotation and Invariance

### 2.1 Contrast: Raw PCA vs. Varimax-Rotated Factor Loadings

| Metric / Concept | Raw PCA Loading Vector | Varimax-Rotated Factor Loading Vector |
| :--- | :--- | :--- |
| **Mathematical Objective** | Maximizes variance sequentially along orthogonal directions. | Maximizes the variance of the squared loadings within columns of $\boldsymbol{\Lambda}$. |
| **Structure / Alignment** | **Global / Dense:** The first component attempts to account for as much total system variance as possible, resulting in dense vectors where most sensors have non-zero coefficients. | **Simple Structure / Sparse:** Drives factor loadings toward $1$, $-1$, or $0$. Each sensor tends to load highly onto only one factor, and each factor isolates a distinct group of sensors. |
| **Physical Interpretation** | Identifies the dominant geometric axes of total data variance; difficult to isolate to independent physical mechanisms. | Isolates **statistically independent physical subsystems** (e.g., mapping Factor 1 exclusively to "vibration subsystems" and Factor 2 exclusively to "thermal subsystems"). |

---

### 2.2 Proof of Invariance Under Orthogonal Rotations
Let $\mathbf{T}$ be an orthogonal transformation matrix such that $\mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}_{k \times k}$. Let the rotated loading matrix be $\boldsymbol{\Lambda}^* = \boldsymbol{\Lambda}\mathbf{T}$.

#### 1. Global Covariance Approximation Invariance
The common-factor component of the global covariance matrix under the rotated loadings is calculated as:
$$\boldsymbol{\Lambda}^* (\boldsymbol{\Lambda}^*)^T = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T = \boldsymbol{\Lambda}(\mathbf{T}\mathbf{T}^T)\boldsymbol{\Lambda}^T$$

Since $\mathbf{T}$ is orthogonal ($\mathbf{T}\mathbf{T}^T = \mathbf{I}$):
$$\boldsymbol{\Lambda}^* (\boldsymbol{\Lambda}^*)^T = \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$$

Thus, the global covariance approximation remains completely unchanged:
$$\mathbf{R} \approx \boldsymbol{\Lambda}^*(\boldsymbol{\Lambda}^*)^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\subsection{\Psi}}$$

#### 2. Communality ($h_j^2$) Invariance
The communality of the $j$-th sensor is the $j$-th diagonal element of the matrix product $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$. Let $\mathbf{\lambda}_j^T$ denote the $j$-th row of $\boldsymbol{\Lambda}$. The rotated row vector is given by $(\mathbf{\lambda}_j^*)^T = \mathbf{\lambda}_j^T \mathbf{T}$.

The rotated communality $(h_j^*)^2$ is the squared Euclidean norm of this row vector:
$$(h_j^*)^2 = \sum_{l=1}^k (\lambda_{jl}^*)^2 = (\mathbf{\lambda}_j^*)^T \mathbf{\lambda}_j^* = (\mathbf{\lambda}_j^T \mathbf{T})(\mathbf{\lambda}_j^T \mathbf{T})^T = \mathbf{\lambda}_j^T (\mathbf{T}\mathbf{T}^T) \mathbf{\lambda}_j$$

Using the property $\mathbf{T}\mathbf{T}^T = \mathbf{I}$:
$$(h_j^*)^2 = \mathbf{\lambda}_j^T \mathbf{I} \mathbf{\lambda}_j = \mathbf{\lambda}_j^T \mathbf{\lambda}_j = h_j^2$$

> **Conclusion:** Orthogonal rotations merely shift the coordinate axes within the latent factor subspace. While this shifts the coordinate values (altering individual entries of $\boldsymbol{\Lambda}$), it preserves the total length of the row vectors (communality) and the total projected variance space (global covariance matrix).

## Part 3: Latent Factor Estimation via Thomson’s Regression Method

### 3.1 Partitioned Covariance Matrix Construction
We construct the joint partitioned random vector:
$$\mathbf{Y}_i = \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix}$$

Because $\mathbb{E}[\mathbf{Z}_i] = \mathbf{0}$ and $\mathbb{E}[\mathbf{F}_i] = \mathbf{0}$, the joint population covariance matrix is block-structured as:
$$\boldsymbol{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \mathbb{E}\left[ \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \begin{bmatrix} \mathbf{Z}_i^T & \mathbf{F}_i^T \end{bmatrix} \right] = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \end{bmatrix} = \begin{bmatrix} \boldsymbol{\Sigma}_{11} & \boldsymbol{\Sigma}_{12} \\ \boldsymbol{\Sigma}_{21} & \boldsymbol{\Sigma}_{22} \end{bmatrix}$$

Let us evaluate each block individually:
1. **Top-Left Block ($\boldsymbol{\Sigma}_{11}$):** By definition, $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{R}$.
2. **Bottom-Right Block ($\boldsymbol{\Sigma}_{22}$):** By definition of the orthogonal common factors, $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$.
3. **Top-Right Block ($\boldsymbol{\Sigma}_{12}$):** Substitute the generative equation $\mathbf{Z}_i = \boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i$:
   $$\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]$$
   Applying $\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}$ and $\mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T] = \mathbf{0}$:
   $$\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbf{I} + \mathbf{0} = \boldsymbol{\Lambda}$$
4. **Bottom-Left Block ($\boldsymbol{\Sigma}_{21}$):** This is the transpose of the top-right block:
   $$\mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] = (\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T])^T = \boldsymbol{\Lambda}^T$$

Assembling the validated blocks into the joint covariance matrix gives:
$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$

---

### 3.2 Derivation of Thomson’s Score Estimator Equation
We match our partitioned system to the Multivariate Normal conditional mean projection theorem. Let:
* $\mathbf{X}_1 = \mathbf{Z}_i \implies \boldsymbol{\mu}_1 = \mathbb{E}[\mathbf{Z}_i] = \mathbf{0}$
* $\mathbf{X}_2 = \mathbf{F}_i \implies \boldsymbol{\mu}_2 = \mathbb{E}[\mathbf{F}_i] = \mathbf{0}$

From the partitioned covariance matrix $\boldsymbol{\Sigma}_{YY}$ derived in Section 3.1, we identify:
$$\boldsymbol{\Sigma}_{11} = \mathbf{R}$$
$$\boldsymbol{\Sigma}_{21} = \boldsymbol{\Lambda}^T$$

The projection identity states:
$$\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \boldsymbol{\mu}_2 + \boldsymbol{\Sigma}_{21} \boldsymbol{\Sigma}_{11}^{-1} (\mathbf{x}_1 - \boldsymbol{\mu}_1)$$

Substituting our specific system matrices and vector realizations ($\mathbf{x}_1 = \mathbf{z}_i$) into the projection identity:
$$\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i] = \mathbf{0} + \boldsymbol{\Lambda}^T \mathbf{R}^{-1} (\mathbf{z}_i - \mathbf{0})$$

Simplifying the expression yields the final score equation for Thomson's regression estimator:
$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$

### Develop a Mathematical Engine & Pre-processing Guardrails


Implement an isolated algorithm to structure and decompose empirical data structures according to the following guidelines:

1. **Z-Score Mapping Layer:** Accept an arbitrary data frame and isolate numeric observation columns. Standardize the observations into standard normal space $Z \in \mathbb{R}^{n \times m}$ applying an unbiased sample standard deviation ($ddof=1$). Build an explicit numerical guardrail that catches zero-variance channels ($\sigma_j = 0$) and caps them to a minimum boundary threshold of $10^{-15}$ to prevent divide-by-zero runtime exceptions.
2. **Dimensionality Constraint Layer:** Let $m$ represent the count of isolated sensor features, and $k$ represent the target count of latent physical factors. Program two validation rule exceptions:
* Reject processing if the operational snapshot count ($n$) is less than or equal to the channel count ($m$).
* Throw an descriptive error if the requested latent space dimension ($k$) is not strictly less than the available feature space dimension ($m$).


3. **Information Decomposition Layer:** Fit a Factor Analysis framework utilizing a Maximum Likelihood or Singular Value Decomposition (SVD) algorithm combined with an orthogonal Varimax rotation matrix ($T$). From the fitted parameters, compute the vector arrays for:
* **Communality Vector ($h^2 \in \mathbb{R}^m$):** The total shared variance captured across each channel.
* **Uniqueness Vector ($\varphi^2 \in \mathbb{R}^m$):** The remaining isolated sensor noise floor.

````python
import numpy as np
import pandas as pd
from scipy.linalg import svd


def factor_analysis_decomposition(
    df: pd.DataFrame, target_factors: int, varimax_max_iter: int = 500, varimax_tol: float = 1e-6
) -> dict:
    """Structures and decomposes empirical data structures using Factor Analysis

    with a Varimax rotation layer.

    Parameters:
    -----------
    df : pd.DataFrame
        The arbitrary input data frame containing mixed observation types.
    target_factors : int (k)
        The target count of latent physical factors to extract.
    varimax_max_iter : int
        Maximum iterations allowed for the orthogonal Varimax rotation.
    varimax_tol : float
        Convergence tolerance for the Varimax rotation.

    Returns:
    --------
    dict: A dictionary containing the standard scores, factor loadings,
          communalities, and uniqueness vectors.
    """
    # -------------------------------------------------------------------------
    # 1. Z-Score Mapping Layer & Isolation
    # -------------------------------------------------------------------------
    # Isolate numeric observation columns
    numeric_df = df.select_dtypes(include=[np.number])
    X = numeric_df.to_numpy(dtype=np.float64)

    n, m = X.shape

    # 2. Dimensionality Constraint Layer
    if n <= m:
        raise ValueError(
            f"Dimensionality Constraint Violation: Operational snapshot count (n={n}) "
            f"must be greater than the channel feature count (m={m})."
        )

    if target_factors >= m:
        raise ValueError(
            f"Dimensionality Constraint Violation: Requested latent space dimension (k={target_factors}) "
            f"must be strictly less than the available feature space dimension (m={m})."
        )

    # Calculate mean and unbiased sample standard deviation (ddof=1)
    means = np.mean(X, axis=0)
    stds = np.std(X, axis=0, ddof=1)

    # Explicit numerical guardrail for zero-variance channels
    stds = np.where(stds == 0, 1e-15, stds)

    # Standardize into standard normal space Z
    Z = (X - means) / stds

    # -------------------------------------------------------------------------
    # 3. Information Decomposition Layer (SVD Framework)
    # -------------------------------------------------------------------------
    # Compute the sample correlation matrix from standard normal space
    R = np.corrcoef(Z, rowvar=False)

    # Decompose correlation matrix using Singular Value Decomposition
    U, S, Vt = svd(R)

    # Extract unrotated factor loadings: L = U_k * sqrt(S_k)
    # Truncate to the requested target latent factor space (k)
    U_k = U[:, :target_factors]
    S_k = np.diag(S[:target_factors])
    L_unrotated = U_k @ np.sqrt(S_k)

    # -------------------------------------------------------------------------
    # Orthogonal Varimax Rotation Layer
    # -------------------------------------------------------------------------
    # Optimization maximizes the variance of the squared loadings across columns
    L_rotated = np.copy(L_unrotated)
    p, q = L_rotated.shape
    T = np.eye(q)  # Initialize orthogonal rotation matrix

    for _ in range(varimax_max_iter):
        T_old = np.copy(T)

        # Kaiser normalization wrapper step: element-wise operations
        Lambda = L_rotated
        d = np.sum(Lambda**2, axis=0)
        # Vectorized Varimax gradient step
        B = (Lambda**3) - (1.0 / p) * Lambda @ np.diag(d)
        # Singular Value Decomposition of the gradient map to find the closest rotation matrix
        U_v, _, Vt_v = svd(Lambda.T @ B)
        T = U_v @ Vt_v

        # Update current rotated loadings
        L_rotated = L_unrotated @ T

        # Check convergence tolerance against the Frobenius norm of difference
        if np.linalg.norm(T - T_old) < varimax_tol:
            break

    # Calculate final Decomposition Vector Arrays
    # Communality (h^2): Total shared variance captured across each channel
    communalities = np.sum(L_rotated**2, axis=1)

    # Uniqueness (phi^2): Remaining isolated sensor noise floor
    uniqueness = 1.0 - communalities

    return {
        "Z_scores": Z,
        "factor_loadings": L_rotated,
        "communality_vector": communalities,
        "uniqueness_vector": uniqueness,
        "rotation_matrix": T,
        "feature_names": list(numeric_df.columns),
    }


# =============================================================================
# Execution Sandbox / Verification Example
# =============================================================================
if __name__ == "__main__":
    # Generate an artificial empirical dataset (Synthetic Sensor Grid)
    np.random.seed(42)
    snapshots = 100
    channels = 5

    # Base underlying physical factors
    factor_1 = np.sin(np.linspace(0, 10, snapshots))
    factor_2 = np.cos(np.linspace(0, 10, snapshots))

    # Construct dataframe with mixed telemetry types (including non-numeric data)
    mock_data = {
        "Timestamp": pd.date_range(start="2026-01-01", periods=snapshots, freq="h"),
        "Sensor_A": 2.5 * factor_1 + np.random.normal(0, 0.2, snapshots),
        "Sensor_B": -1.2 * factor_1 + np.random.normal(0, 0.5, snapshots),
        "Sensor_C": 4.0 * factor_2 + np.random.normal(0, 0.1, snapshots),
        "Sensor_D": 0.5 * factor_2 + np.random.normal(0, 0.9, snapshots),
        "Sensor_Dead": np.ones(snapshots) * 42.0,  # Forces zero-variance channel trigger
        "Status_Flag": ["OPERATIONAL"] * snapshots,  # Categorical string channel
    }
    df_empirical = pd.DataFrame(mock_data)

    print("--- Initializing Empirical Matrix Decomposition ---")
    k_factors = 2

    try:
        metrics = factor_analysis_decomposition(df_empirical, target_factors=k_factors)

        print("\n[✓] Standardized Z-score matrix shape:", metrics["Z_scores"].shape)
        print("\n[✓] Rotated Factor Loadings matrix (m x k):")
        for name, loadings in zip(metrics["feature_names"], metrics["factor_loadings"]):
            print(f"    {name:12}: {loadings}")

        print("\n[✓] Communality Vector (h²):")
        for name, comm in zip(metrics["feature_names"], metrics["communality_vector"]):
            print(f"    {name:12}: {comm:.6f}")

        print("\n[✓] Uniqueness Vector (φ²):")
        for name, uniq in zip(metrics["feature_names"], metrics["uniqueness_vector"]):
            print(f"    {name:12}: {uniq:.6f}")

    except ValueError as e:
        print(f"\n[X] Constraint Exception Tripped: {e}")
````

## 1. How Varimax Rotation Relaxes PCA Rules to Create "Simple Structure"

Traditional Principal Component Analysis (PCA) operates under a strict, unyielding mathematical constraint: each successive principal component must maximize the remaining total variance in the dataset while remaining orthogonal to the previous components. This creates a rigid hierarchy ($\lambda_1 > \lambda_2 > \dots$). Because `PC 1` is forced to capture as much mixed variance as possible across the entire system, it almost always ends up with uninterpretable, cross-coupled coefficients where nearly every sensor has a moderate loading.

Factor Analysis (FA) with a **Varimax rotation** fundamentally changes this objective by prioritizing **interpretability** over raw variance maximization:

* **Orthogonal Coordinate Rotation:** Varimax leaves the underlying factor space intact but physically rotates the factor axes.
* **Maximizing Variance of the Loadings:** Instead of maximizing the variance *of the data projection*, Varimax maximizes the variance of the *squared loadings within each factor*.
* **Driving Coefficients to Extremes:** This mathematical optimization pushes the loading values toward the extremes—either close to $\pm1.0$ or close to $0.0$.

Consequently, the mixed, murky cross-couplings are cleared away. This creates a **"simple structure"** where each sensor aligns heavily with exactly one factor and drops to near-zero on all others. This is precisely why your heatmap shows `Sensor_1` and `Sensor_2` cleanly isolated on `Factor 1` (the primary structural mode), while `Sensor_3` shifts exclusively to `Factor 2` (the secondary operational mode).

* **From $k=2$ to $k=3$:** The curve flattens out abruptly, forming a distinct **"elbow"** at $k=2$. The drop from $k=2$ to $k=3$ is minimal because the remaining variance in the system consists almost entirely of un-correlated, localized sensor noise (such as the massive white noise on `Sensor_4`), rather than shared structural dynamics.

## 2. Why the "Flat Elbow" Identifies $k=2$ as the True Physical Dimensionality

In multi-sensor asset monitoring, the true hidden physical dimensionality is the number of latent forces driving systemic, co-varying changes across multiple channels.

* **The $k=2$ Pivot Point:** The sharp drop before $k=2$ indicates that the first two components are capturing highly structured, shared variance ($\text{Sensor}_1$, $\text{Sensor}_2$, and $\text{Sensor}_3$ coupled to $f_1$ and $f_2$).
* **The Post-$k=2$ Flatline:** The plateau after $k=2$ signifies a transition from modeling **system signal** to modeling **isolated noise**. Because adding a third component yields negligible improvements in reducing systemic error, it proves that there are no more underlying physical mechanisms left to extract. The data-generating process is fundamentally driven by exactly two underlying factors.

## 3. The Danger of Over-fitting: Choosing $k=3$

If you incorrectly set your truncation cutoff to $k=3$ for the automated monitoring framework, you commit an over-fitting error with severe engineering consequences:

* **Infiltrating the Clean Subspace with Noise:** You force the third principal component to absorb the massive, localized instrumentation fault from `Sensor_4`. This un-correlated white noise is now treated as a "clean," normal physical state of the asset.
* **Crippling the Anomaly Detection Engine:** The $Q$ statistic is designed to flag anomalies by monitoring when data spills out of the principal subspace into the residual space. By moving `Sensor_4`'s severe electrical noise *into* the clean subspace ($k=3$), your residual space loses its baseline sensitivity.
* **Operational Blindness:** If `Sensor_4` fails completely or experiences a catastrophic spike, the $Q$ statistic will barely react because that noise has been legitimized by the $k=3$ model. Conversely, the Hotelling's $T^2$ control limit will be perpetually stretched out by this noise, masking real, subtle mechanical deviations occurring on the healthy sensors.

## Operational Trade-Offs: PCA Subspace Monitoring vs. Rotated FA Factor Scores

When deploying a real-time predictive maintenance pipeline for a fleet of identical industrial assets, choosing between a Principal Component Analysis (PCA) framework and a Factor Analysis (FA) framework dictates how resilient your automated alerts will be against instrumentation anomalies.

---

## 1. Comparative Analysis of Monitoring Strategies

### Approach 1: The PCA Strategy ($T^2$ and $Q$ Statistics)
The PCA strategy monitors the asset by splitting the data space into two orthogonal tracking zones: the principal subspace (measured via Hotelling's $T^2$) and the residual/noise subspace (measured via the $Q$ statistic or Squared Prediction Error).

* **How it works:** Hotelling's $T^2$ checks if the overall asset state has drifted away from the baseline coordination of the main components. The $Q$ statistic checks if a sensor has broken out of the model entirely, indicating that the data no longer fits the baseline PCA correlation structure.
* **Operational Vulnerability:** Because PCA mixes both shared and unique variance together during its coordinate calculation, a severe localized anomaly can bleed across *both* statistics simultaneously.

### Approach 2: The FA Strategy (Rotated Latent Factor Scores)
The FA strategy bypasses general variance tracking and monitors the underlying, physical driving forces ($f_1$ and $f_2$) directly via rotated factor scores.

* **How it works:** The real-time tracking focuses purely on the isolated physical phenomena (e.g., `Factor_1` tracking structural load changes, and `Factor_2` tracking operational speed variations).
* **Operational Safeguard:** Each sensor's localized, un-correlated noise floor is segregated before the factor scores are computed.

---

## 2. Robustness Against Sensor Failure: Why FA Wins

**The Factor Analysis (FA) Strategy is significantly more robust** against a single sensor losing calibration or experiencing a sudden electrical short.

The justification lies entirely within how the two engines handle the **Sensor Uniqueness Noise Floor ($\psi^2$)**:

### The FA Defense Mechanism (Using Uniqueness)
In your asset profile, `Sensor_4` possesses a uniqueness value ($\psi^2$) exceeding $98\%$. The Factor Analysis algorithm explicitly extracts and isolates this uniqueness into a diagonal error covariance matrix ($D$) *before* computing the latent factor scores:

$$\Sigma = \Lambda \Lambda^T + \Psi$$

Because the $98\%$ white noise of `Sensor_4` is partitioned into $\Psi$, the mathematical weights used to calculate the scores for `Factor_1` and `Factor_2` will effectively assign a weight near zero to `Sensor_4`.

* **The Result:** If `Sensor_4` experiences a massive electrical short or loses calibration in the field, the spike is safely absorbed by its pre-allocated uniqueness floor. The health scores for your true physical processes (`Factor_1` and `Factor_2`) remain perfectly stable, avoiding a false plant shutdown.

### Why PCA Fails under the Same Scenario
PCA does not possess a uniqueness partitioning mechanism. It treats all variance equally.

* **The Result:** If a sensor loses calibration or shorts out, the massive localized variance surge cannot be contained. In a PCA pipeline, this error will instantly contaminate `PC_1` or `PC_2`, causing Hotelling’s $T^2$ to spike wildly. Concurrently, because the broken sensor violates the baseline correlation structure, the $Q$ statistic will also breach its control limit. The pipeline flags a massive systemic anomaly, forcing operators to hunt for a "ghost" structural failure when it was merely a single loose sensor wire.

| **Root-Cause Diagnostics** | **Moderate:** Requires secondary evaluation via complex asset contribution plots. | **Excellent:** Anomaly aligns instantly with a specific, isolated physical factor heatmap. |
